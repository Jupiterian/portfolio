## Popcorn Hack 2: Sum and Average with Accumulator Pattern

```python
# CODE_RUNNER: Lists Popcorn 2
# Use the accumulator pattern to sum and average a list of numbers.
# The accumulator pattern has three parts:
# 1. Initialize total to 0 BEFORE the loop
# 2. Add each element to total INSIDE the loop
# 3. Use the total AFTER the loop

numbers = [7, 6, 8, 5, 7, 9]

# Initialize accumulator
total = 0

# Add each number to the total
for number in numbers:
    total += number

# Calculate average
print(f"Sum: {total}")

average = total / len(numbers)
print(f"Average: {average}")
```

## Popcorn Hack 3: Frequency Counting and Filtering

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter
# Count how many times each value appears in a list,
# then filter to find which values appear 2 or more times.

seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]

# Step 1: Count frequency of each season
frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1

# Step 2: Filter to find seasons that appear 2 or more times
popular = []
for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

print("Frequency count:")
print(frequency)
print("\nPopular seasons (appear 2+ times):")
print(popular)
```

## Homework: Complete Dataset Analysis

This exercise combines all three lessons:
- Lesson 1: Indexing and list length
- Lesson 2: Sum and average with accumulator
- Lesson 3: Frequency counting and filtering

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter
# Analyze voting data for snack preferences and their prices.

votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]
prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# ========== LESSON 1: Indexing ==========
print("=== LESSON 1: Indexing ===")
print(f"Number of votes: {len(votes)}")
print(f"Last vote: {votes[-1]}")

# ========== LESSON 2: Sum and Average ==========
print("\n=== LESSON 2: Sum and Average ===")
total = 0
for price in prices:
    total += price

average = total / len(prices)
print(f"Total cost: {int(total)}")
print(f"Average price: {average}")

# ========== LESSON 3: Count and Filter ==========
print("\n=== LESSON 3: Frequency Count and Filter ===")

# Count how many votes each snack received
frequency = {}
for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1

# Find snacks that received 3 or more votes
popular = []
for snack in frequency:
    if frequency[snack] >= 3:
        popular.append(snack)

print(f"Vote frequency: {frequency}")
print(f"Popular snacks (3+ votes): {popular}")
```
