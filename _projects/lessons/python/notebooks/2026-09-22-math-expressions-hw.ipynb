## Part 1 — What is a Math Expression?

A math expression combines numbers (or variables) and operators so Python evaluates them into a single new value.

| Operator | Meaning | Description | Usage Example |
|----------|---------|-------------|----------------|
| `+` | Addition | Adds two values together | `10 + 5` evaluates to `15` |
| `-` | Subtraction | Subtracts second from first | `10 - 5` evaluates to `5` |
| `*` | Multiplication | Multiplies two values | `10 * 5` evaluates to `50` |
| `/` | Division | Divides first by second | `10 / 5` evaluates to `2.0` |
| `**` | Exponentiation | Raises first to power of second | `2 ** 3` evaluates to `8` |
| `//` | Floor Division | Divides, rounds down | `10 // 3` evaluates to `3` |
| `%` | Modulus | Returns the remainder | `10 % 3` evaluates to `1` |

## Going Deeper (optional): Reverse Polish Notation

Reverse Polish Notation (RPN) removes the need for parentheses and order-of-operations rules — every expression can be evaluated left to right with a stack.

Example: `5 3 - 2 *` (equivalent to `(5 - 3) * 2`)

| Step | Remaining Input | Operation | Stack |
|------|------------------|-----------|-------|
| 1 | `5 3 - 2 *` | Push 5 | `[5]` |
| 2 | `3 - 2 *` | Push 3 | `[5, 3]` |
| 3 | `- 2 *` | Pop 3, pop 5 → 5 - 3 → push 2 | `[2]` |
| 4 | `2 *` | Push 2 | `[2, 2]` |
| 5 | `*` | Pop 2, pop 2 → 2 * 2 → push 4 | `[4]` |
| 6 | (empty) | Result | `[4]` |

RPN is still used today in some scientific and financial calculators because it's fast to evaluate and never ambiguous.
