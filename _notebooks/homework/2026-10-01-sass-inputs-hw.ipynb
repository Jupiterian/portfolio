## OCS SASS Inputs (PVO)

Source lesson: [sass/inputs](https://pages.opencodingsociety.com/sass/inputs/)

Grammar used: base `ocs__input` plus one size (`small` / `medium` / `large`) and, where a decorative gradient is wanted, `gradient`. HTML defines the field, SASS styles it, and JavaScript would read `.value`.

## Popcorn Hack

The 400px dashed-purple search field becomes `ocs__input large gradient` (one large, gradient-styled input). The made-up `box` class becomes `ocs__input small` (one compact plain input).

````python
%%html
<!-- UI_RUNNER: Inputs Popcorn Base -->
<input type="text" class="ocs__input large gradient" placeholder="Search PVO resources...">
<input type="text" class="ocs__input small" placeholder="Volunteer First Name">
````

## Homework Hack

| Field | Before | After | Why |
|-------|--------|-------|-----|
| Assistance Category | `style="width: 500px; background: linear-gradient(...)"` | `ocs__input large gradient` | the main field, and the gradient now comes from the modifier instead of inline CSS |
| Volunteer First Name | `class="tinybox"` | `ocs__input small` | short value, compact field |
| Volunteer Email | `style="padding: 6px;"` | `ocs__input medium` | standard size; kept `type="email"` so phones show the @ keyboard and the browser can check the format |

For accessibility, each input also gets a visible `<label>` tied to it with `for` / `id` (a placeholder disappears once you start typing, so it isn't a real label), and a `name` so a form submission knows which value is which.

````python
%%html
<!-- UI_RUNNER: Inputs Homework Base -->
<label for="pvo-assistance-category">Assistance category</label>
<input id="pvo-assistance-category" name="assistanceCategory" type="text" class="ocs__input large gradient" placeholder="Assistance Category">

<label for="pvo-volunteer-first-name">Volunteer first name</label>
<input id="pvo-volunteer-first-name" name="volunteerFirstName" type="text" class="ocs__input small" placeholder="Volunteer First Name">

<label for="pvo-volunteer-email">Volunteer email</label>
<input id="pvo-volunteer-email" name="volunteerEmail" type="email" class="ocs__input medium" placeholder="Volunteer Email">
````

**Checklist:** `%%html` and `UI_RUNNER` present. No inline `style` attributes, no made-up classes (`box` and `tinybox` removed). `ocs__input` plus a size modifier on every input, and `gradient` on the decorative field.

**Same interaction in AP CSP pseudocode:**
```
category ← INPUT("Assistance category")
DISPLAY("Searching PVO resources for: " + category)
```
The HTML input is the browser's version of `INPUT()`. The SASS only changes how it looks, not what value it collects.
