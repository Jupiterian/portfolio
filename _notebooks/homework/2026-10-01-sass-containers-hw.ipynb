## Popcorn Hack: Parade Invite Rescue

- `page-wrap` + inline `max-width` → one `ocs__container`
- purple inline `<h1>` → `ocs__section-title` inside an `ocs__card`, plus a one-line `ocs__description`
- inline flex row + `box` divs → `ocs__grid ocs__grid--standard cols-3` with three `ocs__grid-cell`s, the start time as `--accent`
- no `style` attributes or made-up classes left

```python
%%html
<!-- UI_RUNNER: Containers Popcorn - Parade Invite Rescue -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">You're invited to the Poway Scripps Rotary Parade, a morning of floats, music and neighbors on Poway Road.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00 AM</div>
      <div class="ocs__grid-cell">What you get: An invaluable local and cultural experience!</div>
    </div>
  </div>
</div>
```

**Test:** at full width the three cells sit side by side. At phone width the standard grid wraps to fewer columns on its own. Nothing in the markup sets a width, so the container's `max-width` and the grid handle both sizes.

## Homework Hack: Parade Hub Page

Every required piece, in order:

| Requirement | Where |
|-------------|-------|
| exactly one `ocs__container` holding one `ocs__card` | outer two divs |
| title + one-line description | `ocs__section-title`, `ocs__description` |
| `cols-3` standard grid with a `--header`, an `--accent` and a `--muted` cell | the grid |
| table rebuilt with `ocs__table-wrap` + `ocs__table`, header row + two body rows | the table |
| one `ocs__callout` at the end | last element in the card |

The `teal` inline `<h1>`, `wrap` / `tile` classes, `display: flex` and `border='1'` table are all gone.

```python
%%html
<!-- UI_RUNNER: Containers Homework - Poway Rotary Parade Hub -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade</h2>
    <p class="ocs__description">Everything you need for parade day, from where to stand to which floats are confirmed.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Day at a Glance</div>
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00 AM</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Parade floats: line-up begins at 7:30 AM</div>
    </div>
    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <tr><th>Entry</th><th>Status</th></tr>
        <tr><td>Rotary Club float</td><td>Open for volunteers</td></tr>
        <tr><td>Poway High marching band</td><td>Confirmed</td></tr>
      </table>
    </div>
    <div class="ocs__callout">Arrive early: Poway Road closes to traffic at 8:00 AM, and the best viewing spots fill up fast.</div>
  </div>
</div>
```

**Test (narrow and wide):** wide, the header spans the top row and the three detail cells sit beneath it. Narrow, the grid wraps and the table stays inside `ocs__table-wrap`, so it lines up with the card instead of overflowing the page.

**Why the cell types fit:** the header labels the whole grid. The start time is the one detail everyone needs, so it's accented. The float line-up time is for participants, not spectators, so it's muted.
