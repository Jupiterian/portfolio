## Popcorn Hack 1: Fix the Beaker Row

The starter used `style="display: flex;"` and a made-up `box` class. It's now one `ocs__grid`, one standard variant, and an `ocs__grid-cell` per beaker. No `style` attribute or `box` class is left.

## Popcorn Hack 2: Size the Beakers with `fr`

All 4 beakers in one row means **4** `fr` values. The Control Beaker is last and twice as wide, so it gets `2fr` and the others get `1fr`. That's 5 shares in total, so each beaker gets 1/5 of the row and the Control Beaker gets 2/5. This is the one hack where a `style` is allowed.

```python
%%html
<!-- UI_RUNNER: Grids Popcorn 2 -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
```

## MCQ

**MCQ Grids: 4/4 | answers: A,B,C,D**

1. **A standard grid with 4 boxes per row: what do you add?** **A** `cols-4`. It's the same as `1fr 1fr 1fr 1fr`.
2. **`grid-template-columns: 2fr 1fr`?** **B** 2 boxes, the first twice as wide. Two values means two boxes, and 3 shares split 2/3 and 1/3.
3. **Which cell type spans the whole row?** **C** `ocs__grid-cell--header`. (`--wide` only spans 2 columns.)
4. **Which line breaks the rule?** **D** `<div class="my-row" style="display:flex">`. It's a made-up class plus an inline style.

## Homework Hack

**Grid 1** is a standard grid.
