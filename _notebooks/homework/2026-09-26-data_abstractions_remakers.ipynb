## Homework

Refactored the 3.1 prototype into one list. Each element is the stars collected on one level, in level order (level 1 = 12, level 2 = 18, level 3 = 10). Then I add 5 to the second level, append a new level with 7 stars, and total it with Example C's loop.

Prediction: [12, 23, 10, 7], 4, 52

````python
# CODE_RUNNER: UESL 3.2 Homework - grow the level collection
level_stars = [12, 18, 10]

if len(level_stars) >= 2:
    level_stars[1] = level_stars[1] + 5
level_stars.append(7)

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)
print(len(level_stars))
print(total_stars)
````

Actual output: [12, 23, 10, 7], 4, 52. Matches the prediction.

The `if len(level_stars) >= 2` check skips the second-level update when there is no second level, otherwise `level_stars[1]` would crash on a short list.

**Why a list is better than 100 separate variables:** with separate variables I'd need `level_1_stars` through `level_100_stars` (100 lines) and a total line that adds all 100 by name. Adding level 101 would mean writing a new variable AND editing the total line, and it's easy to forget one. With a list, all the stars are one line of data, a new level is one `append`, and the `for` loop doesn't change at all because it goes through however many levels there are.

Also like in 3.1, changing the list doesn't update a total that was already calculated. The loop has to run again (with `total_stars` reset to 0) to get the new total.

## Design Thinking

| Phase | What I Did |
|-------|-----------|
| **Need** | A game maker wants to add or reorder levels without losing track of each level's stars |
| **Goal** | Keep every level's stars in one collection, in level order, that's easy to read, update and add to |
| **Representation considered** | Separate variables for each level (like 3.1) vs. one list + a `for` loop. Went with the list |
| **Prototype** | `level_stars = [12, 18, 10]` with an update, an append and Example C's loop for the total |
| **Actual test** | Ran the main list (got 4 levels, 52 stars) and the empty list (got 0, 0) |
| **Revision** | Running the update on an empty list would crash with an IndexError since there's no `level_stars[1]`, so I added `if len(level_stars) >= 2` before it |
