## Popcorn

### Popcorn Hack 1: Define and call

`location` and `status` are the **parameters** (in the definition). `"Poway High"` and `"all clear"` are the **arguments** (in the call).

```python
# CODE_RUNNER: 3.12 Popcorn 1 - Define report_status and call it twice
def report_status(location, status):
    print(f"Status at {location}: {status}")

report_status("Poway High", "all clear")
report_status("Community Park", "needs assistance")
```

### Popcorn Hack 2: A helper inside a procedure

`process_report` calls the helper `is_priority`, which returns `True` or `False`. The outer procedure uses that result to pick the response.

```python
# CODE_RUNNER: 3.12 Popcorn 2 - Helper procedure called by an outer procedure
def is_priority(incident_type):
    return incident_type == "fire" or incident_type == "medical"

def process_report(location, incident_type):
    if is_priority(incident_type):
        print(f"PRIORITY: Dispatch team to {location} for {incident_type}.")
    else:
        print(f"Routine: Log {incident_type} at {location} for follow-up.")

process_report("Oak Street", "fire")
process_report("Library", "noise complaint")
```

## Homework: Poway Neighborhood Emergency Corps

`check_priority` is the helper (decides the response level). `process_incident` is the outer procedure that calls it. Three calls use different incident types and locations.

```python
# CODE_RUNNER: 3.12 Homework - Emergency response with helper + outer procedure
def check_priority(incident_type):
    if incident_type == "fire" or incident_type == "medical":
        return "HIGH"
    elif incident_type == "flooding":
        return "MEDIUM"
    else:
        return "LOW"

def process_incident(location, incident_type):
    level = check_priority(incident_type)
    print(f"Incident: {incident_type} at {location}")
    print(f"Priority: {level}")
    if level == "HIGH":
        print("Response: Dispatch emergency team immediately.")
    elif level == "MEDIUM":
        print("Response: Send a volunteer crew within the hour.")
    else:
        print("Response: Log it for the next daily review.")
    print()

process_incident("Oak Street", "fire")
process_incident("Poway Library", "flooding")
process_incident("Community Park", "noise complaint")
```
