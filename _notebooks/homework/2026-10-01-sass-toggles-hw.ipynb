## SASS Toggles

Source lesson: [sass/toggles](https://pages.opencodingsociety.com/sass/toggles)

Toggle anatomy: `ocs__toggle` (the `<label>` wrapper) → `ocs__toggle-input` (the hidden checkbox that stores on/off) → `ocs__toggle-track` (the visible switch) → `ocs__toggle-label` (the text). `ocs__toggle--green` sets the checked color.

## Popcorn Hack: Fill in the Blanks

1. The **checkbox** on a toggle keeps track of whether it is on or off.
2. The **slider** is the visible part of the toggle that changes color when it is turned on.
3. The `.switch` class controls the toggle's position, size, padding, and **alignment**.
4. The **border-radius** property makes the slider have rounded edges.

### Find the Bug

The track `<span>` had an empty `class=""`, so the visible switch never rendered. Adding `ocs__toggle-track` fixes it. `aria-hidden="true"` is there because the track is decoration only: screen readers announce the checkbox and its label instead.

```python
%%html
<!-- UI_RUNNER: Toggles Popcorn - find the bug -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="bugfix-toggle" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Enable Notifications</span>
</label>
<p id="bugfix-result"></p>
<script>
(function () {
  // Same check the lesson's runner does: is the track class present now?
  const ok = document.querySelector("#bugfix-toggle + .ocs__toggle-track") !== null;
  document.getElementById("bugfix-result").textContent = ok ? "Toggle works! 🎉" : "Incorrect, almost there!";
})();
</script>
```

### Customize the Toggle

The lesson's color runner, changed from green `#22C55E` to purple `#8B5CF6`. This `<style>` is the lesson's own starter, the one place the lesson asks for a hex edit.

```python
%%html
<!-- UI_RUNNER: Toggles Popcorn - custom color -->
<style>
.custom .ocs__toggle-input:checked + .ocs__toggle-track {
  background: #8B5CF6;
}
</style>
<div class="custom">
  <label class="ocs__toggle">
    <input class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">My Custom Toggle</span>
  </label>
</div>
```

## Homework: Parade-Day Settings Panel

Four toggles, each with its own label and its own job on the page:

| Toggle | What it does when ON |
|--------|---------------------|
| Show parade schedule | reveals the schedule list |
| Show volunteer tips | reveals the tips paragraph |
| Show route map note | reveals the route note |
| Compact view | hides every optional section at once (it overrides the other three) |

A live counter shows how many toggles are on and updates on every flip. The JavaScript only reads `.checked` and sets `.hidden`, so all the styling still comes from the OCS classes.

```python
%%html
<!-- UI_RUNNER: Toggles Homework - parade-day settings panel -->
<div>
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-toggle-schedule" class="ocs__toggle-input" type="checkbox" checked>
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show parade schedule</span>
  </label>
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-toggle-tips" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show volunteer tips</span>
  </label>
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-toggle-route" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show route map note</span>
  </label>
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-toggle-compact" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Compact view (hide everything)</span>
  </label>

  <p><output id="hw-toggle-count" aria-live="polite"></output></p>

  <ul id="hw-schedule">
    <li>7:30 AM: Float line-up</li>
    <li>9:00 AM: Parade starts on Poway Road</li>
    <li>11:00 AM: Community picnic</li>
  </ul>
  <p id="hw-tips">Volunteer tip: wear sunscreen, bring water, and check in at the Rotary tent.</p>
  <p id="hw-route">Route note: the parade runs west on Poway Road from Community Park to Midland Road.</p>
</div>
<script>
(function () {
  // Wrapped in a function so these const names don't collide with other cells on the same page.
  const schedule = document.getElementById("hw-toggle-schedule");
  const tips = document.getElementById("hw-toggle-tips");
  const route = document.getElementById("hw-toggle-route");
  const compact = document.getElementById("hw-toggle-compact");
  const all = [schedule, tips, route, compact];

  function update() {
    // .checked is the whole idea: the checkbox remembers the state, and JS just reads it.
    // Compact view wins over the other three. It's like an outer if around the inner checks.
    document.getElementById("hw-schedule").hidden = compact.checked || !schedule.checked;
    document.getElementById("hw-tips").hidden = compact.checked || !tips.checked;
    document.getElementById("hw-route").hidden = compact.checked || !route.checked;

    // Count how many toggles are on. I loop instead of hardcoding 4 so adding a toggle later still works.
    let onCount = 0;
    all.forEach(function (t) {
      if (t.checked) {
        onCount += 1;
      }
    });
    document.getElementById("hw-toggle-count").textContent =
      "Toggles on: " + onCount + " of " + all.length;
  }

  // "change" fires every time a toggle is flipped, so the page and the counter stay in sync.
  all.forEach(function (t) {
    t.addEventListener("change", update);
  });
  update(); // run once at load so the page matches the starting states
})();
</script>
```

**Tested states:** at load only the schedule is on, so the counter reads "Toggles on: 1 of 4" and only the schedule shows. Turning on tips and route shows all three sections, 3 of 4. Turning on compact view hides all three sections even though their toggles are still on, 4 of 4. Turning everything off gives 0 of 4 and nothing shows.

**What I learned:** the toggle is a styled checkbox. `ocs__toggle-track` only *looks* like a switch, and the real state lives in the hidden `<input>`. That's why `:checked` in the SCSS and `.checked` in JavaScript read the same thing. It's AP CSP selection: the checkbox is a Boolean, and `if (t.checked)` picks a branch.
