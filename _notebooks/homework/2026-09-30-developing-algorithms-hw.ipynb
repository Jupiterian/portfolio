### Popcorn Hack 1: Translate pseudocode to Python

Prediction: 3 + 5 + 2 + 4 = 14, and 14 / 4 = 3.5

````python
# CODE_RUNNER: 3.09 Popcorn 1 - Translate the pseudocode into Python
# Step 1: make the hours list
hours = [3, 5, 2, 4]

# Step 2: start total at 0
total = 0

# Step 3: FOR EACH -> for loop, add each h to total
for h in hours:
    total = total + h

# Step 4: LENGTH(hours) -> len(hours), compute the average
average = total / len(hours)

# Step 5: print total, then average
print(total)
print(average)
````

Actual output: 14 and 3.5. `FOR EACH h IN hours` turns into `for h in hours:`, and `LENGTH` turns into `len`.

### Popcorn Hack 2: Predict, then fix

Prediction (before running the original): **1**. With `count > capacity`, 12 isn't greater than 12, so Self-Defense doesn't get counted and only Fitness (15) does.

Fix: changed `>` to `>=` so a workshop at exactly 12 counts as full.

````python
# CODE_RUNNER: 3.09 Popcorn 2 - Predict, then fix the full-workshop counter
signed_up = [12, 7, 15, 9]   # Self-Defense, Parenting Skills, Fitness, Self-Sufficiency
capacity = 12

full = 0
for count in signed_up:
    if count >= capacity:    # was >, which skipped the workshop at exactly 12
        full = full + 1

print("Full workshops:", full)
# My prediction before running: 1 (the original > misses Self-Defense at 12)
````

Actual output: `Full workshops: 2` after the fix (the original printed 1, same as my prediction). It's a one-character bug that doesn't cause any error, it just gives the wrong count at the boundary.

### Popcorn Hack 3: Modify the counter into a collector

````python
# CODE_RUNNER: 3.09 Popcorn 3 - Modify the counter into a collector
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]     # signed_up[0] goes with workshops[0], and so on
capacity = 12

# collect names instead of counting
open_workshops = []
for i in range(len(workshops)):
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])

# print how many, then the list
print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
````

Actual output: `Workshops with open seats: 2` and `['Parenting Skills', 'Self-Sufficiency']`. Self-Defense is at exactly 12 so `< 12` leaves it out. The loop and the if are the same as a counter, the only thing that changed is `count = count + 1` became `open_workshops.append(...)`.

### Homework Hack: Supply closet restock report

Combines three patterns: **collect** the items below target, **sum** the shortages, and **find the max** shortage, then picks between "Fully stocked!" and the report.

Prediction:

- Toothbrushes 40/30 and Notebooks 30/30 aren't short
- Socks is 25/25, exactly at target, so `25 < 25` is False and it's not short either
- Blankets needs 20 - 8 = 12, Shampoo needs 15 - 5 = 10
- Total 22, most urgent is Blankets (12)

````python
# CODE_RUNNER: 3.09 Homework - Safe Passage Heals supply restock report
items    = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]      # in the closet now
target   = [30, 20, 25, 15, 30]    # what staff want on hand

# Step 1: COLLECT items below target, and how many of each are needed
short_items = []
short_amounts = []
for i in range(len(items)):
    if in_stock[i] < target[i]:
        short_items.append(items[i])
        short_amounts.append(target[i] - in_stock[i])

# Step 2: SUM the shortages with a loop (no sum())
total_needed = 0
for amount in short_amounts:
    total_needed = total_needed + amount

# Step 3: FIND THE MAX shortage (track its position so you know the item's name)
most_urgent = 0
for i in range(len(short_amounts)):
    if short_amounts[i] > short_amounts[most_urgent]:
        most_urgent = i

# Step 4: COMBINE: print "Fully stocked!" if nothing is short, otherwise the report
if len(short_items) == 0:
    print("Fully stocked!")
else:
    print("Restock needed:")
    for i in range(len(short_items)):
        print(f"{short_items[i]}: need {short_amounts[i]}")
    print("Total items needed:", total_needed)
    print(f"Most urgent: {short_items[most_urgent]} (short by {short_amounts[most_urgent]})")
````

Actual output:
```
Restock needed:
Blankets: need 12
Shampoo: need 10
Total items needed: 22
Most urgent: Blankets (short by 12)
```
Matches the expected output exactly.

For the max I track the **position** (`most_urgent`) instead of the amount, since `short_items` and `short_amounts` line up by index, so the position gives me both the name and the number.
